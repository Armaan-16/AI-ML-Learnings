Matplotlib

```python
#matplotlib is a data visualisation technique,it is a python library which is used for data visualisation,primarily it is used to create static and interactive plots.
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Line Plot:
# A graph that uses points connected by lines to show trends,
# changes, or relationships between two variables over a sequence.
x=[1,2,3,4,5,6,7,8]
y=[2,3,5,7,11,13,17,19]

plt.plot(x,y,)
plt.title("line plot")
plt.xlabel('x-axix')
plt.ylabel('y-axix')
plt.show()
```

```python
#line plot
x=[1,2,3,4,5,6,7,8]
y=[2,3,5,7,11,13,17,19]

plt.plot(x,y,'r*-.') # r*-. : red color, star marker, dashed line
plt.title("line plot",color='b')
plt.xlabel('x-axix',color='r')
plt.ylabel('y-axix',color='g')
plt.show()
```

```python
#Bar Plot:
#A graph that uses rectangular bars to compare values across categories.
# Reason for the pattern:
# The height of each bar represents its corresponding value.
# Taller bars indicate larger values, while shorter bars indicate smaller values.
# Different colors help distinguish bars visually.
x=[1,2,3,4,5,6,7,8]
y=[2,3,5,7,11,13,17,19]

a=[1,2,3,4,5,6]
b=[3,4,33,13,6,7]

plt.bar(a,b,label='ex n')

plt.bar(x,y,color=['b','g','r','c','y','m'],width=0.9,label='ex m')
plt.title("bar plot")
plt.xlabel('x-axix')
plt.ylabel('y-axix')
plt.show()
```

```python
# Histogram:
# A graphical representation that shows the frequency distribution
# of numerical data by grouping values into intervals called bins.

# Parameters:
# age              : Data values to be plotted.
# bins             : Defines the intervals (ranges) for grouping data.
# rwidth=0.9       : Sets the relative width of the bars.
# edgecolor='k'    : Sets the bar border color to black.
# linewidth=1.3    : Sets the thickness of the bar borders.

# Reason for the pattern:
# Each bar represents a range of values (bin).
# The height of a bar indicates how many data points fall within that range.
# Taller bars indicate higher frequencies, while shorter bars indicate lower frequencies.

age=[10,20,33,45,37,73,30,26,23,66,69,55,12,4]
bins=[10,20,30,40,50,60,70]

plt.hist(age, bins,
         rwidth=0.9,
         edgecolor='k',
         linewidth=1.3)

plt.title('Histogram')
plt.xlabel('Age')
plt.ylabel('Counts')
plt.show()
```

```python
# Scatter Plot:
# A graphical representation that displays individual data points
# to visualize the relationship, distribution, and correlation
# between two numerical variables.
# Reason for the pattern:
# Each point represents one observation (x, y).
# The arrangement of points helps identify trends, clusters,
# correlations, and outliers in the data.
# An upward pattern indicates a positive relationship,
# while a downward pattern indicates a negative relationship.

x=[10,20,33,45,37,53,30]
y=[10,20,30,40,50,60,70]
plt.scatter(x,y,color='g',s=400,marker='*')
plt.title('scatter plot')
#plt.grid()

plt.show()
```

```python
#scatter plots
# Combined Scatter Plot and Bar Plot:
# This graph overlays a scatter plot on top of a bar plot to
# visualize both individual data points and their corresponding values.
# Reason for the pattern:
# The bars represent the magnitude of each y-value.
# The scatter points highlight the exact data values.
# Using both plots together makes it easier to compare values
# while also emphasizing individual observations.

x=[10,20,33,45,37,53,30]
y=[10,20,30,40,50,60,70]
plt.scatter(x,y,color='g',s=400,marker='*')
plt.title('scatter plot')
plt.bar(x,y,color='g',width=0.05)
plt.scatter(x,y,marker='o')
#plt.grid()

plt.show()
```

```python
# Pie Chart:
# A circular statistical graph used to represent the proportion
# or percentage of different categories in a dataset.
# Reason for the pattern:
# Each slice represents a category.
# The size of a slice is proportional to its value.
# Larger slices indicate a greater share of the total,
# while smaller slices indicate a smaller share.
activities=['mobile','sleep','study','eat']
slicies=[2,7,7,3]
cols=['r','g','b','c']
plt.pie(slicies,labels=activities,colors=cols,autopct='%1.1f%%',startangle=90,counterclock=False)
plt.title('pie chart')
plt.show()
```

```python

activities=['mobile','sleep','study','eat']
slicies=[2,7,7,3]
cols=['r','g','b','c']
plt.pie(slicies,labels=activities,colors=cols,autopct='%1.3f%%',startangle=90,counterclock=False,shadow=True,explode=(0.1,0,0,0))
# autopct='%1.3f%%'
# Displays percentages with 3 decimal places.

# startangle=90
# Starts the first slice at 90 degrees (top of the circle).

# counterclock=False
# Draws the slices in a clockwise direction.

# shadow=True
# Adds a shadow effect below the pie chart for better visualization.

# explode=(0.1,0,0,0)
# Separates the first slice from the pie chart by 10% of the radius,
# highlighting or emphasizing that category.
plt.title('pie chart')
plt.show()
```

```python
#subplopts:
# subplot(rows, columns, position)
# Used to display multiple plots in a single figure.
x=[1,2,3,4]
y=[1,4,9,16]

plt.subplot(1,2,1)
plt.plot(x,y)
plt.title("Line Plot")

plt.subplot(1,2,2)
plt.bar(x,y)
plt.title("Bar Plot")

plt.show()
```

```python
# Multiple Line Plot:
# Used to compare two or more datasets on the same graph.

x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

plt.plot(x,y1,label='Squares')
plt.plot(x,y2,label='Numbers')

plt.legend()
plt.show()
```

```python
# Horizontal Bar Plot:
# Similar to bar plot but bars are horizontal.

students=['A','B','C','D']
marks=[70,85,60,90]

plt.barh(students,marks)

plt.title('Student Marks')
plt.show()
```

```python
# Box Plot:
# Shows median, quartiles, spread and outliers.

data=[12,15,20,21,25,28,30,90]

plt.boxplot(data)

plt.title('Box Plot')
plt.show()
```

```python
# Stem Plot:
# Displays discrete data points with vertical stems.

x=[1,2,3,4,5]
y=[10,20,15,30,25]

plt.stem(x,y)

plt.title('Stem Plot')
plt.show()
```

```python
# Area Plot:
# Shows cumulative values using filled regions.

x=[1,2,3,4,5]
y=[2,4,6,8,10]

plt.fill_between(x,y)

plt.title('Area Plot')
plt.show()
```

```python
# Error Bar Plot:
# Shows uncertainty or variation in data.

x=[1,2,3,4]
y=[10,15,20,25]

plt.errorbar(x,y,yerr=2)

plt.title('Error Bar Plot')
plt.show()
```

```python
# Annotation:
# Used to add text notes on graphs.

x=[1,2,3,4]
y=[10,20,30,40]

plt.plot(x,y)

plt.annotate('Peak',
             xy=(4,40))

plt.show()
```

```python
# Grid:
# Adds horizontal and vertical reference lines.

plt.plot([1,2,3,4],[10,20,15,30])

plt.grid()

plt.show()
```

```python
# Axis Limits:
# Controls visible range of axes.

x=[1,2,3,4,5]
y=[10,20,30,40,50]

plt.plot(x,y)

plt.xlim(0,6)
plt.ylim(0,60)

plt.show()
```

```python
# Save Figure:
# Saves graph as an image file.

plt.plot([1,2,3],[4,5,6])

plt.savefig('graph.png')

plt.show()
```

```python
# Tick Rotation:
# Rotates axis labels for better readability.

names=['Armaan','Rahul','Aman','Priya']
marks=[90,85,88,92]

plt.bar(names,marks)

plt.xticks(rotation=45)

plt.show()
```

```python
# Log Scale:
# Used when values vary over a large range.

x=[1,10,100,1000]
y=[2,20,200,2000]

plt.plot(x,y)

plt.xscale('log')

plt.show()
```

```python
# Style:
# Changes the appearance of plots.

plt.style.use('ggplot')

plt.plot([1,2,3,4],[10,20,15,30])

plt.show()
```
