<h1>PANDAS LIBRARY</H1>

Installing and importing Pandas

```python
#pandas is a library used for data manipulation and analysis. It provides data structures and functions needed to manipulate structured data, such as tabular data in the form of DataFrames.
#it is of two types 1) series(single column of data)
#                   2) dataframe
import pandas as pd
import numpy as np
```

1) Series creation

```python
#creating series
s=pd.Series([1,2,3,4,5])
print(s)
type(s)
```

```python
s=pd.Series(['a','b','c','d','e'])
print(s)
```

```python
s=pd.Series(['a','b','c','d','e'],index =[1,2,3,11,5]) # we can set index of objects according to us
print(s)
```

```python
print(s[11]) # indexing
print(s[1:4]) #slicing (by default we have to give the general series of index while slicing)
```

2) Conversion of different datatypes to series 

```python
#tuple to series
t= pd.Series((13,5,3,5)) 
print(t)
```

```python
#set to series
s=pd.Series({2,3,4,4,5,})
s# set can't be used as series it is unordered 
```

```python
#dictonary to series
d=pd.Series({'a':1,'b':2,'c':3})
print(d) #key becomes index position and value becomes the object
d=pd.Series({'a':1,'b':2,'c':3},index=['e','f','a','b'])
print(d)
d['b']# indexing
d['a':'b':1] #slicing
```

3)Dataframes creation

```python
#creating dataframe
d={'name':['ram','hari','sita','gita'],'roll':[101,102,103,104],'course':['AIML','PYTHON','DSA','C']} #dataframes are created using dictionary
print(d)
df=pd.DataFrame(d)
print(df)
```

```python
print(df.name[2]) #indexing syntax:- var.colname[index]
print(df.iloc[::1]) #this is slicing in dataframes
print (df.iloc[2::])
print(df.iloc[::,2])
```

```python
print(df.head()) # it will give us the first 5 rows of dataframe by default and if we want to give the number of rows we want to see then we can give the number in the head function like this df.head(3) it will give us the first 3 rows of dataframe
print(df.tail()) # it will give us the last 5 rows of dataframe by default and if we want to give the number of rows we want to see then we can give the number in the tail function like this df.tail(3) it will give us the last 3 rows of dataframe 
```

```python
df.sample(3) # it will give us a random row from the dataframe and if we want to give the number of rows we want to see then we can give the number in the sample function like this df.sample(3) it will give us 3 random rows from the dataframe
```

```python
df=df.rename(columns={'name':'Name','roll':'Roll','course':'Course'}) # renaming the columns of dataframe
print(df)
```

```python
df=df.set_index('Roll') # setting the index of dataframe to name column
print(df)
#df=df.set_index('Course',inplace=True) # setting the index of dataframe to course column and inplace is used to make the changes in the original dataframe
#print(df)
```
