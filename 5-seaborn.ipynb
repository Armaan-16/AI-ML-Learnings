```python
#seaborn is a python library based on matplotlib which provides a high level interface for drawing attractive and informative statistical graphics. It is built on top of matplotlib and closely integrated with pandas data structures. Seaborn provides a variety of functions for visualizing data, including scatter plots, line plots, bar plots, histograms, and more. It also includes features for customizing the appearance of plots, such as color palettes and themes. Seaborn is widely used in data analysis and machine learning for exploring and visualizing data.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print(sns.__version__)
```

```python
iris=pd.read_csv(r'C:\Users\mishr\Downloads\iris\iris.data',header=None)
iris.columns=['sepal_length','sepal_width','petal_length','petal_width','flowers']
iris
```

```python
# Distribution Plot: A graphical representation used to visualize how data values are distributed across a dataset, showing their frequency and spread.
sns.displot(iris.sepal_width)
plt.show()
```

```python
# Histogram Plot: A graphical representation that displays the frequency distribution of numerical data using bars.
sns.histplot(iris.sepal_width)
plt.show()
```

```python
# KDE Plot: A smooth curve used to visualize the probability density and distribution pattern of numerical data.
sns.kdeplot(iris.petal_length, fill=True)
plt.show()
```

```python
pd.value_counts(iris.petal_length)
```

```python
# hatch='*' adds a star-pattern fill to the shaded area of the plot for better visual distinction.
sns.kdeplot(iris.petal_length, fill=True, hatch='*')
plt.show()
```

```python
# color='red' sets the color of the distribution plot to red.
sns.displot(iris.petal_length, color='red')
plt.show()
```

```python
sns.displot(iris.petal_length[iris['class']=='Iris-setosa'])
sns.displot(iris.petal_length[iris['class']=='Iris-versocolor'])
sns.displot(iris.petal_length[iris['class']=='Iris-virginica'])
plt.show
```

```python
sns.kdeplot(iris.petal_length[iris['flowers']=='Iris-setosa'], label='Setosa')
sns.kdeplot(iris.petal_length[iris['flowers']=='Iris-versicolor'], label='Versicolor')
sns.kdeplot(iris.petal_length[iris['flowers']=='Iris-virginica'], label='Virginica')

plt.legend() #plt.legend() displays a legend on the plot, helping identify different data series using their labels.
plt.show()
```

```python
sns.histplot(data=iris) #this compare all numerical columns in the dataset and plot them together.
plt.show()
```

```python
sns.histplot(data=iris,fill=False,hatch='*')
plt.show()
```

```python
# Box Plot (Box-and-Whisker Plot):
# A graphical representation used to visualize the distribution, spread,
# central tendency, and outliers of numerical data.

# Parts of a Box Plot:
# 1. Minimum (Lower Whisker)  : Smallest non-outlier value.
# 2. First Quartile (Q1)      : 25th percentile of the data.
# 3. Median (Q2)              : Middle value of the dataset.
# 4. Third Quartile (Q3)      : 75th percentile of the data.
# 5. Interquartile Range(IQR) : Q3 - Q1, represents the middle 50% of data.
# 6. Whiskers                 : Lines extending to minimum and maximum non-outlier values.
# 7. Outliers                 : Unusually high or low values lying outside the whiskers.

# Advantages:
# - Shows data distribution and spread.
# - Helps identify outliers.
# - Useful for comparing multiple groups.
# - Provides a quick statistical summary of data.
sns.boxplot(x=iris['flowers'],y=iris.petal_width,hue=iris.flowers)
plt.show()
```

```python
# Boxen Plot (Letter Value Plot):
# An enhanced version of a box plot used to visualize the distribution
# of large datasets by showing more quantiles and detailed data spread.
sns.boxenplot(x=iris.flowers,y=iris.petal_width,palette='cool')
plt.show()
```

```python
# Violin Plot:
# A statistical plot that combines a box plot and a KDE (Kernel Density Estimation)
# plot to show the distribution, spread, density, and probability of data.
# The pattern in a violin plot represents the probability density of the data.
# Wider regions indicate more data points concentrated at those values,
# while narrower regions indicate fewer data points.
sns.violinplot(x=iris.flowers, y=iris.petal_width, palette='cool')
plt.show()
```

```python
# Swarm Plot:
# A categorical scatter plot that displays individual data points
# without overlapping, helping visualize the distribution of data.
# The pattern in a swarm plot represents the density and spread of data points.
# Wider regions indicate a higher concentration of observations,
# while narrower regions indicate fewer observations.
sns.swarmplot(x=iris.flowers, y=iris.petal_width, palette='cool')
plt.show()
```

```python
# Strip Plot:
# A categorical scatter plot that displays individual data points
# along a categorical axis, helping visualize data distribution.
# Pattern in a strip plot represents the concentration and spread of data points.
# Dense regions indicate higher frequency of observations, while sparse regions
# indicate lower frequency.
sns.stripplot(x=iris.flowers, y=iris.petal_width, palette='cool')
plt.show()
```

```python
# Bar Plot:
# A graphical representation that uses rectangular bars to compare
# the average values of a numerical variable across different categories.
# Height of bar : Represents the mean (average) value.
# Error bars    : Show variability/confidence interval.
# Categories    : Displayed on the x-axis.
# Comparison    : Helps compare values across groups.
sns.barplot(x=iris.flowers, y=iris.petal_width, palette='viridis')
plt.show()
```

```python
# Point Plot:
# A statistical plot that displays the estimated central tendency
# (usually the mean) of a numerical variable for different categories.
# The points represent the mean value of each category, and the lines
# connecting them help visualize trends or differences between categories.
# The vertical lines (error bars) show the variability or confidence
# interval around the mean.

sns.pointplot(x=iris.flowers, y=iris.petal_width, palette='rocket')
plt.show()
```

```python
sns.pointplot(x=iris.flowers,y=iris.petal_width)
```

```python
# Scatter Plot:
# A graphical plot that displays individual data points to show
# the relationship or distribution between variables.
# Each dot represents one observation in the dataset.
# Clusters of points indicate areas where many observations occur,
# while gaps indicate fewer observations. The overall pattern helps
# identify distribution, grouping, and potential outliers.

sns.scatterplot(x=iris.flowers,
                y=iris.petal_width,
                hue=iris.flowers, #Assigns different colors to different flower categories
                palette='rocket')
plt.show()
```

```python
# Joint Plot:
# A statistical plot that combines a scatter plot with the
# distribution plots (histograms/KDE) of the individual variables.

# Purpose:
# Used to visualize the relationship between two numerical variables
# along with their individual distributions.

# Reason for the pattern:
# The central scatter plot shows how data points are related,
# while the plots on the top and right margins show the distribution
# of each variable separately. Clusters indicate concentrations of data,
# and the overall pattern can reveal trends, correlations, or outliers.

sns.jointplot(data=iris)
plt.show()
```

```python
# KDE Joint Plot:
# A statistical plot that combines a bivariate KDE (Kernel Density Estimation)
# plot with the individual KDE distributions of each variable.
# The contour curves and shaded regions represent the density of data points.
# Areas with darker or denser contours indicate a higher concentration of observations,
# while lighter or wider-spaced contours indicate fewer observations.
# The marginal KDE plots show the distribution of each variable separately.


sns.jointplot(data=iris,kind='kde')
```

```python
# Count Plot:
# A categorical plot used to display the frequency (count)
# of observations in each category.
# Reason for the pattern:
# The height of each bar represents the number of observations
# in that category. Taller bars indicate more occurrences,
# while shorter bars indicate fewer occurrences.

sns.countplot(x=iris.flowers, palette='hot')
plt.show()
```

```python
sns.countplot(x=iris.petal_length,palette='hot')
```

```python
# Figure Size:
# figsize=(20,15) sets the width and height of the figure in inches,
# making the plot larger and easier to read.

# Count Plot:
# Displays the frequency (count) of each unique petal_length value.

# Reason for the pattern:
# Each bar represents a unique petal_length value.
# Taller bars indicate values that occur more frequently,
# while shorter bars indicate values that occur less frequently.
# The pattern shows the distribution of petal_length frequencies in the dataset.

plt.figure(figsize=(20,15))
sns.countplot(x=iris.petal_length, palette='hot')
plt.show()
```

```python
# Pair Plot:
# A visualization that displays pairwise relationships between
# numerical variables in a dataset using scatter plots and
# distribution plots.

# Reason for the pattern:
# The diagonal plots show the distribution of individual variables.
# The off-diagonal plots show relationships between pairs of variables.
# Clusters, trends, and separations in the plots help identify
# correlations, patterns, and potential outliers.

sns.pairplot(iris)
plt.show()
```

```python
# Pair Plot with Hue:
# Displays pairwise relationships between numerical variables and
# uses different colors to represent different categories/classes.
sns.pairplot(iris,hue='flowers')
```
