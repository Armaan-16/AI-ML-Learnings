```python
#operators:
#Operators are special symbols that perform specific operations on operands (variables and values) and return a result.
#TYPES OF OPERATORS:
#        1.ARITHMETIC OPERATORS: +, -, *, /, //, %, **
# arithmetic operators are used to perform mathematical operations on numeric operands.
a=13
b=3
print("a+b =", a+b)
print("a-b =", a-b)
print("a*b =", a*b)
print("a/b =", a/b)
print("a//b =", a//b)
print("a%b =", a%b)
print("a**b =", a**b)

#        2.ASSIGNMENT OPERATORS: =, +=, -=, *=, /=, //=, %=, **=
# assignment operators are used to assign values to variables and perform operations on them in a concise way.
x=5
print("x =", x)
print("x += 3", x) # it is used to add a value to the variable and assign the result back to the variable. It is equivalent to x = x + 3.
print("x -= 2", x) # it is used to subtract a value from the variable and assign the result back to the variable. It is equivalent to x = x - 2.
print("x *= 4", x) # it is used to multiply the variable by a value and assign the result back to the variable. It is equivalent to x = x * 4.
print("x /= 2", x) # it is used to divide the variable by a value and assign the result back to the variable. It is equivalent to x = x / 2.
print("x //= 3", x) # it is used to perform floor division on the variable by a value and assign the result back to the variable. It is equivalent to x = x // 3.
print("x %= 4", x) # it is used to perform modulus operation on the variable by a value and assign the result back to the variable. It is equivalent to x = x % 4.
print("x **= 2", x) # it is used to perform exponentiation on the variable by a value and assign the result back to the variable. It is equivalent to x = x ** 2.

#        3.COMPARISON/RELATIONAL OPERATORS: ==, !=, >, <, >=, <=
# comparison operators are used to compare two values and return a boolean result (True or False) based on the comparison.
a=10
b=20
print("a:", a)
print("b:", b)
print("a == b:", a == b) # it is used to check if the values of two operands are equal. It returns True if they are equal, otherwise it returns False.
print("a != b:", a != b) # it is used to check if the values of two operands are not equal. It returns True if they are not equal, otherwise it returns False.
print("a > b:", a > b) # it is used to check if the value of the left operand is greater than the value of the right operand. It returns True if it is greater, otherwise it returns False.
print("a < b:", a < b) # it is used to check if the value of the left operand is less than the value of the right operand. It returns True if it is less, otherwise it returns False.
print("a >= b:", a >= b) # it is used to check if the value of the left operand is greater than or equal to the value of the right operand. It returns True if it is greater than or equal, otherwise it returns False.
print("a <= b:", a <= b) # it is used to check if the value of the left operand is less than or equal to the value of the right operand. It returns True if it is less than or equal, otherwise it returns False.

#        4.LOGICAL OPERATORS: and, or, not
# logical operators are used to combine multiple boolean expressions and return a boolean result based on the logical relationship between the expressions.
a = True
b = False
print("a and b:", a and b) # it is used to perform a logical AND operation between two boolean values. It returns True if both operands are True, otherwise it returns False.
print("a or b:", a or b) # it is used to perform a logical OR operation between two boolean values. It returns True if at least one of the operands is True, otherwise it returns False.
print("not a:", not a) # it is used to perform a logical NOT operation on a boolean value. It returns True if the operand is False, and it returns False if the operand is True.

#        5.BITWISE OPERATORS: &, |, ^, ~, <<, >>
#        6.IDENTITY OPERATORS: is, is not
#        The "is" operator is used to check if two variables refer to the same object in memory. It returns True if both variables point to the same object, and False otherwise.
a = [1, 2, 3]
b = a
print("a is b:", a is b) # it will return True because both a and b refer to the same list object in memory.
c = [1, 2, 3]
print(id(a)) # it is used to get the memory address of the object that a refers to.
print(id(c)) # it is used to get the memory address of the object that c refers to.
print("a is c:", a is c) # it will return False because a and c refer to different list objects in memory, even though they have the same content.
m=256
n=256
print("m is n:", m is n) # it will return True because small integers (-5 to 256) are cached by Python and refer to the same memory location.
p=257 
q=257 
print("p is q:", p is q) # it will return False because integers outside the range of -5 to 256 are not cached and refer to different memory locations.
x=(4)
y=(4)
print("x is y:", x is y) # it will return True because small integers are cached by Python and refer to the same memory location.
e={1}
f={1}
print("e is f:", e is f) # it will return False because sets are mutable and each set object is stored at a different memory location, even if they have the same content.
u='hello'
v='hello'
print("u is v:", u is v) # it will return True because string literals are interned by Python and refer to the same memory location if they have the same content.
c='hello world'
d='hello' + ' world'
k=''.join(['hello', ' world'])
print("c is d:", c is d) # it will return True because string literals are interned by Python and refer to the same memory location if they have the same content.
print("c is k:", c is k) # it may return True or False depending on the implementation of Python, as string concatenation at runtime may not always result in the same memory location being used for the resulting string.
#        7.MEMBERSHIP OPERATORS: in, not in
# The "in" operator is used to check if a value exists within a sequence (such as a string, list, tuple, set, or dictionary). It returns True if the value is found in the sequence, and False otherwise.
l=[1, 2, 3, 4, 5]
print("3 in l:", 3 in l) # it will return True because the value 3 is present in the list l.
print("6 in l:", 6 in l) # it will return False because the value 6 is not present in the list l.
s = "hello world"
print("'o' in s:", 'o' in s) # it will return True because the character 'o' is present in the string s.
print("'z' in s:", 'z' in s) # it will return False because the character 'z' is not present in the string s.
d = {"Name": "Ram", "Age": 20, "Roll No": 21}
print("'Name' in d:", 'Name' in d) # it will return True because the key 'Name' is present in the dictionary d.
print("'Ram' in d:", 'Ram' in d) # it will return False because the value 'Ram' is not a key in the dictionary d. To check for values in a dictionary, you would use 'Ram' in d.values(), which would return True.
```
