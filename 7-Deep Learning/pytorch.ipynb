```python
import torch
import time

# 1. Verify GPU Availability
print("Checking hardware...")
if torch.cuda.is_available():
    device = torch.device("cuda")
    print(f"✅ GPU detected: {torch.cuda.get_device_name(0)}")
else:
    device = torch.device("cpu")
    print("❌ GPU not detected. Defaulting to CPU.")

# 2. Benchmark Matrix Multiplication
print("\nRunning benchmark (Matrix Multiplication)...")
matrix_size = 5000

# --- CPU Time ---
start_time = time.time()
cpu_matrix_a = torch.randn(matrix_size, matrix_size)
cpu_matrix_b = torch.randn(matrix_size, matrix_size)
cpu_result = torch.matmul(cpu_matrix_a, cpu_matrix_b)
cpu_time = time.time() - start_time
print(f"⏱️ CPU Time: {cpu_time:.4f} seconds")

# --- GPU Time ---
if torch.cuda.is_available():
    start_time = time.time()
    # Send matrices straight to the GPU VRAM
    gpu_matrix_a = torch.randn(matrix_size, matrix_size).to(device)
    gpu_matrix_b = torch.randn(matrix_size, matrix_size).to(device)
    gpu_result = torch.matmul(gpu_matrix_a, gpu_matrix_b)
    
    # Wait for GPU to finish computing before stopping the clock
    torch.cuda.synchronize(device)
    gpu_time = time.time() - start_time
    print(f"⚡ GPU Time: {gpu_time:.4f} seconds")
    print(f"🚀 Speedup Factor: {cpu_time / gpu_time:.1f}x faster!")
```
