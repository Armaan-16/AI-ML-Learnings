```python
plt.imshow(img[::25,::25,::-1])
plt.show()
# First ::25
# Selects every 25th row.
# Reduces the image height.

# Second ::25
# Selects every 25th column.
# Reduces the image width.

# Third ::-1
# Reverses the color channels.
# Converts BGR → RGB when image is loaded using OpenCV.

# Shape change example:
# Original: (1080, 1920, 3)
# New:      (44, 77, 3)

# This operation is called image downsampling.
# It reduces image size by keeping every 25th pixel.
```
