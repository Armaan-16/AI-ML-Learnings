```python
cap=cv2.VideoCapture(0)
if cap.isOpened():
    ret,frame=cap.read()
    print(ret)
    print(frame)
else:
    ret=False
Image=cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
plt.imshow(Image)
plt.title("Webcam")
plt.xticks([])
plt.yticks([])
plt.show()
cap.release()
```
