```python
#mathematical operations using array
a=np.array([25,73,47])
b=np.array([43,33,53])
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
print(np.floor_divide(a,b))
print(np.mod(a,b))
print(np.divmod(a,b))
print(np.log(a))
print(np.sqrt(b))
print(np.cbrt(a))
print(np.power(a,b))
print(np.sin(a))
print(np.cos(b))
print(np.tan(a))
print(1/np.tan(a))
```

```python
#statistical functions
a=np.array([25,73,47,25])
print(np.max(a))
print(np.min(a))
print(np.argmax(a)) #gives the index value of max value of an array
print(np.argmin(a)) #gives the index value of min value of an array
print(np.sort(a))
print(np.argsort(a))
print(np.flip(a))
#print(flip(np.sort(a)))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.var(a))
print(np.cumsum(a))
print(np.cumprod(a))
print(np.repeat(a,3))
print(np.tile(a,4))
print(np.where(a==25))
print(np.round(38.4))
print(np.round(38.6))
print(np.ceil(38.4)) #gives the higher round off value
print(np.trunc(38.9)) #gives the lower round off value
```

```python
#vector mathematics
a=np.array([[1,2],[1,4]])
b=np.array([[3,7],[5,6],])
print('a=',a)
print('b=',b)

print('dot product=',np.dot(a,b))
print(a@b) #shortcut for dot product
print('cross product=',np.cross(a,b))
print(np.transpose(a))
print(a.T) #shortcut for transpose
```

```python
a=np.array([25,16,28,71,90]) #1D array
print(a)
print(a[2]) #indexing is possible in all types of array
print(a[0:4:2]) # slicing is possible in all types of array
```

```python
a=np.random.randint(1,25,(5,5))
print(a)
print(a[0,2]) #indexing syntax:- var[row,col]
print(a[4,3])
print(a[3,::])
print(a[3:4:1,::])
print(a[3:4:1,1:4:1]) #var[row indexing,col indexing]
print('slicing',a[2:5:2,0:5:2]) #var[staring row: ending row+1: skip, starting col: ending col+1: skip]
print(a[2]) #prints the row 
np.where(a==3)
```

```python
a=np.random.randint(1,25,(5,5))
b=np.random.randint(1,20,(4,4))
c=np.random.randint(1,20,(4,4))
print('a=',a)                                      
print('b=',b)                            # axis 0  |
print('c=',c)                            #         |
# when vertically joined its in axis 0 (default)   | 
# when horizontally joined its in axis 1           |___________ axis 1
# when rows are same they are added vertically  # /
# when cols are same they are added horzontally #/  axis 2
print(np.concatenate(b,c))                     #/
print(np.hstack(a,b))
print(np.hstack(a,b))
print(np.vstack(a,b))

```
