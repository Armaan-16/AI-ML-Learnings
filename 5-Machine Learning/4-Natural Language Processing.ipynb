Steps Involved in NLP/Text Preprossing :<pre>
1)Tokenization- It is a process in which we convert big sentences into tokens of words,or a paragraph into tokens of single sentences.It is of two types 1) sent tonkeniser 2)word tokeniser 
2)Stemming-It is used to find the root or base words from spellings.(detail,details,deatailings)
3)Lemmatization-It is similar to stemmimg it is also used to find root words or base words from the different form of the words(go,went,gone).Output of lemmatization is a proper word.
4)POS Tags:It indicates how a word function in meaning as well as gramatically used within a sentence.(The cat killed the rat.)
5)Named Entity Recognition- It is used to identify weather it is a place, person, location etc.
6)Chunking- picking individual piece of information and grouping it into a bigger pieces is known as chunking.

```python
from sklearn.feature_extraction.text import TfidfVectorizer,TfidfTransformer
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import MultinomialNB
```

```python
model = Pipeline([
    ('vector', TfidfVectorizer()),
    ('tfidf', TfidfTransformer()),
    ('algo', MultinomialNB())
])
```

```python
model.fit(news_train.data,news_train.target)
```

```python
def predict_category(s,tr=news_train,m=model):
    pred=m.predict([s])
    print(pred)
    return tr.target_names[pred[0]]
```

```python
predict_category('audi is better than hyundai')
```

```python
predict_category('the space shuttle is ready for launch')
```

```python
predict_category('he is our president')
```
