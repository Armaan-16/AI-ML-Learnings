```python
iris=pd.read_csv(r"C:\Users\mishr\Downloads\iris\iris.data",header=None)
iris
```

```python
iris.columns=['length','sepal_width','petal_length','petal_width','Flowers' ]
iris
```

```python
#set the input data
ip=iris.drop('Flowers',axis=1)
iris
```

```python
#select output data
op=iris.Flowers
op
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    ip,
    op,
    train_size=0.8,
    random_state=3000
)
```

```python

from sklearn.preprocessing import StandardScaler

sc = StandardScaler() 

x_train = sc.fit_transform(x_train) 

x_test = sc.transform(x_test) 
```

```python
from sklearn.preprocessing import MinMaxScaler

mm = MinMaxScaler()

x_train = mm.fit_transform(x_train)
x_test = mm.transform(x_test)
print(type(x_train))
print(type(x_test))
print(x_test)
```

```python
#ML algorithm used to group similar data points into clusters
from sklearn.cluster import KMeans
km=KMeans(n_clusters=3)#means the algorithm will divide the data into 3 groups (clusters).
km.fit(ip)#Trains the K-Means model on the input data ip.
# During training:
# Randomly selects 3 centroids.
# Assigns each point to the nearest centroid.
# Recalculates centroids.
# Repeats until centroids stop changing.
```

```python
predkm=km.predict(ip)#predicts on the basis on given data
predkm
```

```python
centroid=km.cluster_centers_ #Returns the final centroids found by the K-Means algorithm.
centroid
```

```python
iris['Prediction']=predkm
iris
```

```python
sns.scatterplot(x=iris.petal_length,y=iris.petal_width,c=predkm)
```

```python
sns.scatterplot(x=iris.petal_width,y=iris.petal_length,c=predkm)
sns.scatterplot(x=centroid[::,0],y=centroid[::,2],c='black',s=200)
plt.show()
```
