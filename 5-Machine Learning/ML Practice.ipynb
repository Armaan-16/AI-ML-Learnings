```python
iris=pd.read_csv(r"C:\Users\mishr\Downloads\iris\iris.data",header=None)
iris
```

```python
iris.columns=['length','sepal_width','petal_length','petal_width','Flowers' ]
iris
```

```python
iris.isnull().sum()
```

```python
ip=iris.drop('Flowers',axis=1)
iris
```

```python
op=iris.Flowers
op
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    ip,
    op,
    train_size=0.8,
    random_state=3000
)
```

```python

from sklearn.preprocessing import StandardScaler

sc = StandardScaler() 

x_train = sc.fit_transform(x_train) 

x_test = sc.transform(x_test) 
```

```python
from sklearn.preprocessing import MinMaxScaler

mm = MinMaxScaler()

x_train = mm.fit_transform(x_train)
x_test = mm.transform(x_test)
print(type(x_train))
print(type(x_test))
print(x_test)
```

Logistics Regression

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()

lr.fit(x_train, y_train)

predlr = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score

ac = accuracy_score(y_test, predlr)

rc = recall_score(y_test, predlr, average='macro')

pr = precision_score(y_test, predlr, average='macro')

f1 = f1_score(y_test, predlr, average='macro')

print("Accuracy =", ac)
print("Recall =", rc)
print("Precision =", pr)
print("F1 Score =", f1)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm=confusion_matrix(y_test, predlr)
cm
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
plt.show
```

```python
from sklearn.cluster import KMeans
km=KMeans(n_clusters=3)
km.fit(ip)
```

```python
predkm=km.predict(ip)
predkm
```

```python
centroid=km.cluster_centers_
centroid
```

```python
iris['Prediction']=predkm
iris
```

```python
sns.scatterplot(x=iris.petal_length,y=iris.petal_width,c=predkm)
```

```python
sns.scatterplot(x=iris.petal_width,y=iris.petal_length,c=predkm)
sns.scatterplot(x=centroid[::,0],y=centroid[::,2],c='black',s=200)
plt.show()
```
